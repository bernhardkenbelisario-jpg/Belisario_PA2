## A. Reproducible Normalization Problem

```python
import numpy as np
```

```python
np.random.seed(2112)
X = np.random.randint(10, 101, size=(5, 5))

x_bar = np.mean(X) 
sigma = np.std(X)

X_normalized = (X - x_bar) / sigma
```

```python
print("X:\n", X)
```

```python
print("X_normalized:\n", X_normalized)
```

```python
print("Mean of X_normalized:", np.mean(X_normalized))
```

```python
print("Standard Deviation of X_normalized:", np.std(X_normalized))
```

```python
np.save("X_normalized.npy", X_normalized)
```

## B. Cubes Divisible By 4 Problem

```python
C = (np.arange(1, 101) ** 3).reshape((10, 10))
div_by_4 = C[C % 4 == 0]
```

```python
print("Shape of C:", C.shape)
```

```python
print("div_by_4 array:\n", div_by_4)
```

```python
print("Number of selected elements:", div_by_4.size)
```

```python
np.save("div_by_4.npy", div_by_4)
```

## C. Above-Mean Squares Problem

```python
S = (np.arange(1, 37) ** 2).reshape((6, 6))
S_mean = np.mean(S)
above_mean = S[S > S_mean]
```

```python
print("S:\n", S)
```

```python
print("S_mean:", S_mean)
```

```python
print("above_mean array:\n", above_mean)
```
